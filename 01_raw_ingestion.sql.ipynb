```python

# 1. ვქმნი raw სქემას
spark.sql("CREATE SCHEMA IF NOT EXISTS getdata.raw")


# 2. CSV ფაილის წაკითხვა პირდაპირ GitHub-ის Raw URL-იდან
import pandas as pd

github_raw_url = "https://raw.githubusercontent.com/zzannaa/TBC-Final-Project---Luka-Zanadze/refs/heads/main/vehicles.csv"

# Pandas-ით წაკითხვა და Spark DataFrame-ად გარდაქმნა
df_pd = pd.read_csv(github_raw_url)
df_spark = spark.createDataFrame(df_pd)


# 3. ვინახავ Delta Table-ად getdata.raw სქემაში
df_spark.write.format("delta").mode("overwrite").saveAsTable("getdata.raw.craigslist_used_cars")
```
